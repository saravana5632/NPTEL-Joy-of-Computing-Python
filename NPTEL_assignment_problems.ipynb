```python
# WEEK-2 PROGRAMMING ASSIGNMENT-1
# Discount Calculator for TechMart
# TechMart is running a 10% discount sale on all products. When a customer purchases an item, the cashier enters the original price into the billing system. The system calculates the final price after applying the discount and displays it to the customer. (output with one decimal place printed - not a hard constraint to handle this. program will still work)
#
# Write a Python program that:
# (a) Accepts the original price of an item as input.
# (b) Converts the entered value into a number.
# (c) Calculates the final price after applying a 10% discount.
# (d) Prints the discounted price.

original_price = int(input())
discounted_price = original_price - (original_price * 0.1)
print(discounted_price)
```

```python
# WEEK-2 PROGRAMMING ASSIGNMENT-2
# Library Book Filter
# A school library maintains a list of the number of pages in newly arrived books. To help younger students, the librarian wants to identify books that have more than 300 pages, as these are recommended only for senior students. There is at least one book in given input that has more than 300 pages.

# Write a Python program that:
# (a) Accepts a space-separated list of integers representing the number of pages in each book.
# (b) Identifies all books with more than 300 pages.
# (c) Prints the page counts of these books, separated by spaces.

lst = list(map(int , input().split()))
for i in lst:
    if i >300:
        print(i)
```

```python
#WEEK-2 PROGRAMMING ASSIGNMENT-3

# Even Number Counter
# A teacher is conducting a classroom activity on number patterns. Students are asked to enter a list of numbers, and the computer counts how many of them are even.

# Write a Python program that:
# (a) Accepts a space-separated list of integers.
# (b) Counts how many numbers are even.
# (c) Prints the count.

lst = list(map(int ,input().split()))
count = 0
for i in lst:
    if i%2 == 0:
        count = count + 1
print(count)
```

```python
# WEEK-3 PROGRAMMING ASSIGNMENT-3
# A courier company records package weights (in kg).

# (a) Accept a space-separated list of positive real numbers.
# (b) Compute the average package weight.
# (c) Count how many package weights are greater than the average.

lst = list(map(float , input().split()))
avg = sum(lst)/len(lst)
count = 0
for i in lst:
    if i<avg:
        count+=1
print(count)
```

```python
# WEEK-4 PROGRAMMING ASSIGNMENT-2
# A cybersecurity application generates numeric authentication codes. Before using a code, the software checks whether it is a prime number.

# Write a Python program that:
# Accepts a positive integer n.
# Prints True if n is a prime number; otherwise, prints False.

n = int(input())
isPrime = True
if n < 2:
    isPrime = False
for i in range(2, n//2+1):
    if n%i == 0:
        isPrime = False
print(isPrime)
```

```python
#  WEEK-5 PROGRAMMING ASSIGNMENT-3
# Q.
# A sequence of integers is given. Find all numbers that occur exactly two times.

# Write a function that:
# (a) Accepts a space-separated sequence of integers.
# (b) Returns all numbers that occur exactly two times.
# (c) The numbers should be returned in increasing order.
# (d) If no number occurs exactly two times, returns -1.

lst = list(map(int , input().split()))
freq = {}
for i in lst:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
res = []
for i in freq:
    if freq[i]==2:
        res.append(i)
if len(res)==0:
        print(-1)
res.sort()           # sorting to print increasing value
for i in res:
    print(i , end = " ")
```

```python
#  WEEK-6 PROGRAMMING ASSIGNMENT-1
# A cybersecurity company evaluates password strength by counting the number of alphabetic characters in a password. Due to memory constraints in a secure device, the counting must be performed recursively.

# Write a recursive function that:
# (a) Accepts a string s as input.
# (b) Returns an integer representing the number of alphabetic characters (A-Z and a-z) in the string.
# (c) Does not use loops.

def counter(s):
    if s=="":
        return 0
    if s[0].isalpha():
        return 1 + counter(s[1:])
    else: 
        return counter(s[1:])
        
s=input()
print(counter(s))
```

```python
#  WEEK-6 PROGRAMMING ASSIGNMENT-2
# Q.
# A warehouse records the boxes currently stored in its inventory as a list of box IDs. Implement a program using recursion to determine the total number of boxes in the inventory.

#Write a program that:
# (a) Accepts space-separated integers as input, where each integer represents a box ID.
# (b) Uses recursion to count the number of boxes and displays the count as an integer.
# (c) Does not use loops or the len() function.

def counter(lst):
    if lst == []:
        return 0
    else:
        return 1 + counter(lst[1:])
lst = list(map(int , input().split()))
print(counter(lst))
```

```python
#  WEEK-6 PROGRAMMING ASSIGNMENT-3
# Q.
# A weather station records daily temperatures as a list of values. Implement a program using recursion to determine the highest temperature recorded.

# Write a program that:
# (a) Accepts space-separated integers as input, where each integer represents a recorded temperature.
# (b) Uses recursion to find and display the highest temperature as an integer.
# (c) Does not use loops or the max() function.

def find_max(temp):
    if len(temp)==1:
        return temp[0]
    high = find_max(temp[1:])
    if temp[0]>high:
        return temp[0]
    return high

temp = list(map(int, input().split()))
find_max(temp)
```
